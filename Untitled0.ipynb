Example 2: Sentiment Analysis

```python
def sentiment_analysis(text):
    if "good" in text.lower():
        return "Positive 😊"
    elif "bad" in text.lower():
        return "Negative 😞"
    else:
        return "Neutral 😐"

demo = gr.Interface(fn=sentiment_analysis, inputs="text", outputs="label")
demo.launch()
```
