Example 2: Sentiment Analysis

```python
def sentiment_analysis(text):
    if "good" in text.lower():
        return "Positive 😊"
    elif "bad" in text.lower():
        return "Negative 😞"
    else:
        return "Neutral 😐"

demo = gr.Interface(fn=sentiment_analysis, inputs="text", outputs="label")
demo.launch()
```

Example 3: Image Classifier (Digits Dataset)

```python
import gradio as gr
from sklearn.datasets import load_digits
from sklearn.linear_model import LogisticRegression

digits = load_digits()
X, y = digits.data, digits.target
model = LogisticRegression(max_iter=10000).fit(X, y)

def predict_digit(img):
    img = img / 16.0
    img = img.reshape(1, -1)
    return {str(i): float(p) for i, p in enumerate(model.predict_proba(img)[0])}

demo = gr.Interface(fn=predict_digit, inputs="image", outputs="label", live=True)
demo.launch()
```

Example 4: Chatbot

```python
def chatbot(input_text, history=[]):
    history.append(("User: " + input_text, "Bot: I heard you say " + input_text))
    return history, history

chat = gr.ChatInterface(fn=chatbot)
chat.launch()
```

Example 6: HuggingFace Sentiment Pipeline

```python
from transformers import pipeline
classifier = pipeline("sentiment-analysis")

def hf_sentiment(text):
    return classifier(text)[0]

demo = gr.Interface(fn=hf_sentiment, inputs="text", outputs="label")
demo.launch()
```

Example 7: ML App (Iris Classification)

```python
%%writefile ml_demo.py
import streamlit as st
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

iris = load_iris()
X, y = iris.data, iris.target
clf = RandomForestClassifier().fit(X, y)

st.title("🌼 Iris Flower Prediction")

sepal_length = st.slider("Sepal length", 4.0, 8.0, 5.0)
sepal_width  = st.slider("Sepal width", 2.0, 5.0, 3.0)
petal_length = st.slider("Petal length", 1.0, 7.0, 4.0)
petal_width  = st.slider("Petal width", 0.1, 3.0, 1.0)

if st.button("Predict"):
    pred = clf.predict([[sepal_length, sepal_width, petal_length, petal_width]])[0]
    st.success(f"Predicted Class: {iris.target_names[pred]}")
```
